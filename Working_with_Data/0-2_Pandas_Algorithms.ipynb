```python
import numpy as np
import pandas as pd
```

# **Basic pd.Series arithmetic**

```python
ser = pd.Series(range(3), dtype=pd.Int64Dtype())

ser
```

```python
ser + 42
```

```python
ser - 42
```

```python
ser * 42
```

```python
ser / 42
```

```python
ser2 = pd.Series(range(10, 13), dtype=pd.Int64Dtype())

ser + ser2
```

```python
ser1 = pd.Series([1., 2., 3.], dtype=pd.Float64Dtype())
ser2 = pd.Series([4., pd.NA, 6.], dtype=pd.Float64Dtype())

ser.add(ser2)
```

```python
ser1.add(ser2, fill_value=0)
```

```python
ser1 = pd.Series(range(3), dtype=pd.Int64Dtype())
ser2 = pd.Series(range(3), dtype=pd.Int64Dtype())

ser1 + ser2
```

```python
ser3 = pd.Series([2, 4], dtype=pd.Int64Dtype())

ser1 + ser3
```

```python
ser4 = pd.Series([2, 4, 8], index=[1, 2, 3], dtype=pd.Int64Dtype())

ser1 + ser4
```

```python
ser5 = pd.Series([2, 4, 8], index=[0, 1, 1], dtype=pd.Int64Dtype())

ser1 + ser5
```

```python
import timeit
```

```python
func = lambda: pd.Series(range(10000), dtype=pd.Int64Dtype()) + pd.Series(range(10000), dtype=pd.Int64Dtype())
timeit.timeit(func, number=10_000)
```

```python
func = lambda: pd.Series(range(10_000), dtype=pd.Int64Dtype()).add(pd.Series(range(10_000), dtype=pd.Int64Dtype()))
timeit.timeit(func, number=10_000)
```

# **Basic pd.DataFrame arithmetic**

```python
np.random.seed(42)

df = pd.DataFrame(
    np.random.randn(3, 3),
    columns=["col1", "col2", "col3"],
    index=["row1", "row2", "row3"],
).convert_dtypes(dtype_backend="numpy_nullable")

df
```

```python
df + 1
```

```python
df * 2
```

```python
ser = pd.Series(
    [20, 10, 0],
    index=["col1", "col2", "col3"],
    dtype=pd.Int64Dtype(),
)

ser
```

```python
df + ser
```

```python
ser = pd.Series(
    [20, 10, 0, 42],
    index=["col1", "col2", "col3", "new_column"],
    dtype=pd.Int64Dtype(),
)

ser + df
```

```python
ser = pd.Series(
    [20, 10, 0],
    index=["row1", "row2", "row3"],
    dtype=pd.Int64Dtype(),
)

ser
```

```python
df + ser
```

```python
ser = pd.Series(
    [20, 10, 0, 42],
    index=["row1", "row2", "row3", "row4"],
    dtype=pd.Int64Dtype(),
)

ser
```

```python
df
```

```python
df.add(ser, axis=0) # match up the row label
```

```python
df + df
```

```python
np.random.seed(42)

df2 = pd.DataFrame(np.random.randn(3, 3)).convert_dtypes(dtype_backend="numpy_nullable")

df2
```

```python
df + df2
```
